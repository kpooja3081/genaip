```python
folder = r"C:\Users\mecha\OneDrive\Desktop\Pooja_personal\krish\data"
story = []
for filename in os.listdir(folder):
    with open(os.path.join(folder, filename), encoding = 'unicode_escape') as f:
        story.append(f.read())
    raw_sent = sent_tokenize(corpus)
    for i in raw_sent:
        story.append(simple_preprocess(i))
```

```python
custom_model = gensim.models.Word2Vec(window=5, min_count=1, vector_size=4)
```

```python
custom_model.build_vocab(story)
```

```python
custom_model.corpus_count
```

```python
custom_model.train(story, total_examples=custom_model.corpus_count, epochs=10)
```
