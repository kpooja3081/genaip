```python
def encode_image_to_base64(image_path):
    with open(image_path, "rb") as image_file:
        return base64.b64encode(image_file.read()).decode("utf-8")
```

```python
from langchain_core.messages import HumanMessage

# Create vision model
vision_llm = ChatOpenAI(model="gpt-4o-mini", temperature=0.9)

# Extract text from image using vision model
message = vision_llm.invoke([
    HumanMessage(
        content=[
            {
                "type": "image_url",
                "image_url": {
                    "url": f"data:image/png;base64,{base_64_image}"
                }
            },
            {
                "type": "text",
                "text": "Extract all text from this image. Return only the text content."
            }
        ]
    )
])

extracted_text = message.content
print("Extracted Text:")
print(extracted_text)
```

```python
from openai import OpenAI

# Initialize OpenAI client
client = OpenAI()

# Generate image from text
text_prompt = "A beautiful sunset over mountains with a lake in the foreground"

response = client.images.generate(
    model="dall-e-3",
    prompt=text_prompt,
    size="1024x1024",
    quality="standard",
    n=1
)

image_url = response.data[0].url
print(f"Generated image URL: {image_url}")
print(f"Prompt used: {text_prompt}")
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

load_dotenv()
hf_token = os.environ.get("HF_TOKEN")

# Create embeddings model (sentence embeddings)
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

# Test with some text
text = "What is artificial intelligence?"
vector = embeddings.embed_query(text)

print(f"Text: {text}")
print(f"Embedding dimension: {len(vector)}")
print(f"First 5 values: {vector[:5]}")
```
